```python
import os       #serve per leggere le variabili d'ambiente
from pathlib import Path        #per gestire i percorsi diversi dalla cartella corrente
from dotenv import load_dotenv      #importiamo solo la funziona load perchè ci serve solamente quella
import pandas as pd      

load_dotenv(Path("..") / ".env")        #carica le variabili d'ambiente dal file .env nella cartella superiore (.env)      

chiave = os.getenv("ENTSOE_API_KEY")        #legge la variabile d'ambiente
print(chiave[:4] + "...")             #stampa solo i primi 4 caratteri della chiave
```

```python
from entsoe import EntsoePandasClient


client = EntsoePandasClient(api_key=chiave)     #leggendo la documentazione entsoe su github, il client richiede la chiave come parametro
```

```python
start = pd.Timestamp('20261101', tz='Europe/Rome')      #il client richiede uno start e un end come timestamp, quindi li creo con pandas. Il formato è YYYYMMDD come indicato nella documentazione entose
end = pd.Timestamp('20261108', tz='Europe/Rome')
country_code = 'IT_NORD'        #nella documentazione sono indicati i codici dei paesi

prezzi = client.query_day_ahead_prices(country_code, start=start, end=end)      #assegno il risultato alla viariabile prezzi

print(type(prezzi))      
print(len(prezzi))
print(prezzi.index.tz)
prezzi.head(8)
```

```python
maschera = prezzi.index < end       #creo una maschera per togliere l'ultimo giorno false dal momento che per il calcolo della media oraria sfaserebbe i risultati.

prezzi_orari = prezzi[maschera].resample('h').mean()        #assegno a prezzi_orari tutti i true della maschera e faccio il resample (con h per indicare media oraria) e calcolo la media dei 4 quarti
```
